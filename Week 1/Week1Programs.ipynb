```python
#Correct Code
#Code 1: Prime Checker

from math import sqrt 

testNumber = int(input('Enter a number: '))
 
def primeChecker(number):
    n = sqrt(number)    
    for i in range(2, int(n)+1, 1):
        if number % i == 0:
            return 'Not Prime'
    return 'Prime'

if testNumber < 2:
    print('Not a prime number')
else:
    print(primeChecker(testNumber))
```
